```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import pandas as pd
import time

def scrape_box_office_data():
    # URL of the page to scrape
    url = "https://www.boxofficemojo.com/year/world/2024/"
    
    # Set up Selenium WebDriver
    options = webdriver.ChromeOptions()
    options.add_argument("--headless")  # Run in headless mode (no GUI)
    driver = webdriver.Chrome(options=options)

    try:
        driver.get(url)
        time.sleep(5)  # Allow time for the page to fully load

        # Locate the table containing the data
        table = driver.find_element(By.CLASS_NAME, "imdb-scroll-table")
        rows = table.find_elements(By.TAG_NAME, "tr")

        data = []
        for row in rows[1:102]:  # Ensure it includes the top 100 movies
            cells = row.find_elements(By.TAG_NAME, "td")
            if len(cells) >= 7:
                data.append({
                    "Rank": cells[0].text.strip(),
                    "Release Group": cells[1].text.strip(),
                    "Worldwide Gross": cells[2].text.strip(),
                    "Domestic Gross": cells[3].text.strip(),
                    "Domestic %": cells[4].text.strip(),
                    "Foreign Gross": cells[5].text.strip(),
                    "Foreign %": cells[6].text.strip()
                })

        # Convert to DataFrame and save to CSV
        df = pd.DataFrame(data)
        df.to_csv("box_office_top_100_2024.csv", index=False)
        print("Data successfully saved to box_office_top_100_2024.csv")

    except Exception as e:
        print(f"Error occurred: {e}")

    finally:
        driver.quit()

if __name__ == "__main__":
    scrape_box_office_data()
```
